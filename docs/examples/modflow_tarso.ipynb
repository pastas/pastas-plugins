# Single cell MODFLOW model - DRN Drain

This notebook demonstrates how a single cell MODFLOW model with a DRN (drain) package and modified storage can be used as a stressmodel in Pastas to model threshold groundwater behavior.

## Packages

```python
import logging
from pathlib import Path

import flopy
import pandas as pd
import pastas as ps
from pastas.timer import SolveTimer

from pastas_plugins import modflow as ppmf

bindir = Path("bin")
if not (bindir / "mf6").exists():
    bindir.mkdir(parents=True, exist_ok=True)
    flopy.utils.get_modflow(bindir, repo="modflow6-nightly-build")
```

```python
ppmf.set_log_level(logging.ERROR)

ps.set_use_cache(True)
```

## Load Data

```python
df = pd.read_csv("data/aftopping.csv", index_col=0, parse_dates=True)
head = df["B28H1808_2"].dropna()
prec = df["Precipitation"]
evap = df["Evaporation"]
```

## TARSO Model

```python
ml_tarso = ps.Model(head, constant=False)
ml_tarso.add_stressmodel(
    ps.TarsoModel(
        prec=prec, evap=evap, rfunc=ps.Exponential(), name="exp", oseries=head
    )
)
ml_tarso.solve()
ml_tarso.plot(figsize=(10, 3));
```

We generate a model with ModflowDrnSto, which adds a drain at a certain threshold-value, and modifies the storage coefficient above this level. This model implements the same physics as TarsoModel, and calibrating it independently gives a similar simulation.

```python
# Generate a Model with ModflowDrnSto
ml_mds = ps.Model(head, name="mfdrnsto", constant=False)
mfml_mds = ppmf.ModflowModel(
    model=ml_mds,
    exe_name=bindir / "mf6",
    sim_ws=Path("mfdrnsto"),
    silent=True,
    iconvert=1,
    newtonoptions=["NEWTON"],
    solver_kwargs={
        "complexity": "SIMPLE",
        "outer_dvclose": 1e-2,
        "inner_dvclose": 1e-2,
        "rcloserecord": 1e-1,
        "linear_acceleration": "BICGSTAB",
    },
)
rch = ppmf.ModflowRch(prec, evap)
drn = ppmf.ModflowDrn()
ghb = ppmf.ModflowGhb()
mfml_mds.add_modflow_package([ghb, rch, drn])
ml_mds.add_stressmodel(mfml_mds)
```

```python
ml_mds.parameters
```

```python
# take parameters from the model with the TarsoModel
p_tarso = ml_tarso.parameters["optimal"]

# DIS
ml_mds.set_parameter("DIS_H", initial=p_tarso["exp_d1"] - p_tarso["exp_d0"])

# RCH
ml_mds.set_parameter("RCH_f", initial=p_tarso["exp_f"])

# GHB
ml_mds.set_parameter("constant_d", initial=p_tarso["exp_d0"])
ml_mds.set_parameter("GHB_C", initial=1 / p_tarso["exp_A0"])

# STO
sy = p_tarso["exp_a0"] / p_tarso["exp_A0"]
ss = p_tarso["exp_a1"] / p_tarso["exp_A1"]
ml_mds.set_parameter("STO_Ss", initial=ss)
ml_mds.set_parameter("STO_Sy", initial=sy)

# DRN
ml_mds.set_parameter("DRN_C", initial=1 / p_tarso["exp_A1"])

ml_mds.parameters
```

```python
ax = ml_mds.plot(figsize=(10, 3));
```

```python
with SolveTimer() as t:
    ml_mds.solve(diff_step=1e-4, callback=t.timer)
```

```python
ax = ml_mds.plot(figsize=(10, 3));
```

We can plot the difference between the two calibrated models.

```python
# plot the difference between both models
h1 = ml_tarso.simulate()
h2 = ml_mds.simulate()
(h1 - h2).dropna().plot(figsize=(10, 3));
```

## ThresholdTransform Model

```python
ml_tt = ps.Model(head)
ml_tt.add_stressmodel(ps.RechargeModel(prec=prec, evap=evap, rfunc=ps.Exponential()))
ml_tt.add_transform(ps.ThresholdTransform())
ml_tt.solve()
ml_tt.plot(figsize=(10, 3));
```

We try to model this system with ModflowSto. However, this does not give us the same simulation.

```python
ml_ms = ps.Model(head, name="mfsto", constant=False)
mfml_ms = ppmf.ModflowModel(
    model=ml_ms,
    exe_name=bindir / "mf6",
    sim_ws=Path("mfsto"),
    silent=True,
    iconvert=0,
)
# ghb = ppmf.ModflowGhb()
rch = ppmf.ModflowRch(prec, evap)
mfml_ms.add_modflow_package([rch])
ml_ms.add_stressmodel(mfml_ms)
with SolveTimer() as t:
    ml_ms.solve(diff_step=1e-4, callback=t.timer)
ax = ml_ms.plot(figsize=(10, 3));
```

With ModflowDrnSto we come closer, but we do not get the same result as the Pastas Model with the ThresholdTransform. It is not possible to model the ThresholdTransform in a one-cell Modflow model. ThresholdTransform calculates the discharged water using the non-transformed simulation. However, in the Modflow model, the calculation of the discharge is calculated using the simulation that is affected by the larger storage coefficient.

```python
# Generate a Model with ModflowDrnSto
ml_mds2 = ps.Model(head, name="mfdrnsto2", constant=False)
mfml_mds2 = ppmf.ModflowModel(
    model=ml_mds2,
    exe_name=bindir / "mf6",
    sim_ws=Path("mfdrnsto"),
    silent=True,
    iconvert=1,
)
rch = ppmf.ModflowRch(prec, evap)
ghb = ppmf.ModflowGhb()
drn = ppmf.ModflowDrn()
mfml_mds2.add_modflow_package([ghb, rch, drn])
ml_mds2.add_stressmodel(mfml_mds2)
with SolveTimer() as t:
    ml_mds2.solve(diff_step=1e-4, callback=t.timer)
ax = ml_mds2.plot(figsize=(10, 3))
```
